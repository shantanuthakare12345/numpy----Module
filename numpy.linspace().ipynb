```python
import numpy as np
a=np.linspace(0,10)
print("Content of a\n{}  type of a={}".format(a,type(a)))
```

```python
len(a)
```

```python
import numpy as np
a=np.linspace(0,10,endpoint=False)
print("Content of a\n{}  type of a={}".format(a,type(a)))
```

```python
len(a)
```

```python
import numpy as np
a=np.linspace(0,10,num=10)
print("Content of a\n{}  type of a={}".format(a,type(a)))
```

```python
len(a)
```

```python
import numpy as np
a=np.linspace(0,10,num=10,endpoint=False)
print("Content of a\n{}  type od a={}".format(a,type(a)))
```

```python
len(a)
```

```python
import numpy as np
a=np.linspace(0,10,num=10,retstep=True)
print("Content of a={}  Type of a={}".format(a[0],type(a[0])))
print("-"*40)
print("Return Step of a={}  Type of a={}".format(a[1],type(a[1])))
```

```python
import numpy as np
a,res=np.linspace(0,10,num=10,retstep=True)
print("Content of a={}  Type of a={}".format(a,type(a)))
print("-"*40)
print("Return Step of a={}  Type of a={}".format(res,type(res)))
```

```python
import numpy as np
a,res=np.linspace(100,120,num=15,retstep=True)
print("Content of a={}  Type of a={}".format(a,type(a)))
print("-"*40)
print("Return Step of a={}  Type of a={}".format(res,type(res)))
```
