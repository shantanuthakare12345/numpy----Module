```python
import numpy as np
val = np.random.randint(100,200,5)
print("Random vals={}  type={}".format(val,type(val)))
print("Dim of val={}  Shape={}".format(val.ndim,val.shape))
```

```python
import numpy as np
val = np.random.randint(1000,2000,10)
print("Random vals={}  type={}".format(val,type(val)))
print("Dim of val={}  Shape={}".format(val.ndim,val.shape))
```

```python
import numpy as np
val = np.random.randint(100,200,(2,2))
print("Random vals=\n{}  type={}".format(val,type(val)))
print("Dim of val={}  Shape={}".format(val.ndim,val.shape))
```

```python
import numpy as np
val = np.random.randint(1000,2000,(3,4))
print("Random vals=\n{}  type={}".format(val,type(val)))
print("Dim of val={}  Shape={}".format(val.ndim,val.shape))
```

```python
print("Number of Rows=",val.shape[0])
print("Number of Clos=",val.shape[1])
```

```python
import numpy as np
val = np.random.randint(1000,2000,(2,3,3))
print("Random vals=\n{}  type={}".format(val,type(val)))
print("Dim of val={}  Shape={}".format(val.ndim,val.shape))
```

```python
print("Number of Marices=",val.shape[0])
print("Number of Rows=",val.shape[1])
print("Number of Clos=",val.shape[2])
```
