```python
#Syntax: varname=numpy.random.randn(value)
#gives number of random floating point values depends on 'value'----+ve or -ve
```

```python
val=r.randn(6)
print("Rnadom Floating value={}   type={}".format(val,type(val)))
```

```python
val=r.randn(6)
print("Rnadom Floating value={}   type={}".format(val,type(val)))
print("Dim={}  shape={}".format(val.ndim,val.shape))
```

```python
#Syntax: varname=numpy.random.randn(rows,cols)
#gives number of random floating point values depends on 'Rows x Cols'----+ve or -ve
```

```python
val=r.randn(3,3)
print("Rnadom Floating value=\n{}   type={}".format(val,type(val)))
print("Dim={}  Shape={}".format(val.ndim,val.shape))
```
