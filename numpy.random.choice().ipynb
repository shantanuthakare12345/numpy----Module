```python
#genreates 5 random integer values
import numpy.random as r
vals=r.randint(10,30,size=10)
print("Random valuse={}".format(vals))
#select one value form list of random samples
cv=r.choice(vals)
print("Random choice value=",cv)
```

```python
#genreates 5 random integer values
import numpy.random as r
vals=r.randint(10,30,size=10)
print("Random valuse={}".format(vals))
#select one value form list of random samples
cv=r.choice(vals,5)
print("Random choice value={}  \ntype={}".format(cv,type(cv)))
```

```python
upalps=[chr(val) for val in range(65,91)]
lwalps=[chr(val) for val in range(97,123)]
allalphs=upalps+lwalps
#Select 3 Alphabets from list object as random samples
cv=r.choice(allalphs,3)
print("Random choice Value={}  Type={}".format(cv,type(cv)))
```

```python
upalps=[chr(val) for val in range(65,91)]
lwalps=[chr(val) for val in range(97,123)]
allalphs=upalps+lwalps
#Select 5 Alphabets from list object as random samples
cv=r.choice(allalphs,5)
print("Random choice Value={}  Type={}".format("".join(cv),type(cv)))
```

```python
upalps=[chr(val) for val in range(65,91)]
lwalps=[chr(val) for val in range(97,123)]
allalphs=upalps+lwalps
#Select 5 Alphabets from list object as random samples
cv=r.choice(allalphs,5)
s=""
for v in cv:
    s=s+v
print("Random choice Value={}  Type={}".format("".join(s),type(s)))  
```
