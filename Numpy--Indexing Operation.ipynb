```python
#Numpy--Indexing Operation
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
```

```python
#Indexing Operation on 1D
```

```python
#Get 30 Element from 1dArray
print(a[2])
```

```python
#Get 90 Element from 1dArray with +ve Index
print(a[len(a)-1])
```

```python
#Get 90 Element from 1dArray with +ve Index
print(a[-1])
```

```python
#Get 90 Element from 1dArray with +ve Index
print(a[a.size-1])
```

```python
print(a)
```

```python
#Get 90 Element from 1dArray with -ve Index
print(a[-1])
```

```python
#Get 10 Element from 1dArray with +ve Index
print(a[-a.size])
```

```python
#Indexing Operation on 2D
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(9,)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#Get an Element 50 from 2DArray
print(a[1,1])
```

```python
#Indexing Operation on 3D
import numpy as np
a=np.arange(3*3*3).reshape(3,3,3)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#Get First Matrix From 3D
print(a[0])
```

```python
#Get First Matrix From 3D
print(a[1])
```

```python
#Get First Matrix From 3D
print(a[-1])
```
