```python
#Re-arrange the Element Of ndarray by using permutation()
import numpy.random as r
a=r.randint(10,21,size=8)
print("Original Content Of ndarray a before permutation()=",a)
ra=r.permutation(a)
print("Original Content Of ndarray a after permutation()=",a)
print("Re-arranged Element of ndarray ra=\n{}   type={}  id={}".format(ra,type(ra),id(ra)))
```

```python
#Re-arrange the Element Of ndarray by using permutation()
import numpy.random as r
a=r.randint(10,21,size=8)
print("Original Content Of ndarray a before permutation()=",a)
for i in range(1,len(a)+1):
    ra=r.permutation(a)
    print("Time:{}-Re-arranged Element of ndarray ra={}   \ntype={}  id={}".format(i,ra,type(ra),id(ra)))
```
