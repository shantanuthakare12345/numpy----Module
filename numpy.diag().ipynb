```python
#Syntax:  varname=numpy.diag(ndarray,k)
```

```python
import numpy as np
a=np.arange(9).reshape(3,3)
print("Matrix-A")
print(a)
print("Shape of Matrix-A=",a.shape)
print("Dim=",a.ndim)
```

```python
#Get the Priciple Diagnol Elements
pd=np.diag(a,0)
print("Principle Diag={}  Type={}".format(pd,type(pd)))
```

```python
#Get the Lower Principle Diagnol Elements
print("Matrix-A")
print(a)
pd=np.diag(a,-1)
print("Principle Diag={}  Type={}".format(pd,type(pd)))
```

```python
#Get the Above Principle Diagnol Elements
print("Matrix-A")
print(a)
pd=np.diag(a,k=1)
print("Principle Diag={}  Type={}".format(pd,type(pd)))
```
