```python
#Creating an Object of ndarray by using array()
import numpy as np
a=10
print("Content of a={} Type a={}".format(a,type(a)))
a1=np.array(a)
print("Content of a={} Type a={}".format(a,type(a)))
print("Dimension ndarray={}".format(a1.ndim))
print("Shape of ndarray={}".format(a1.shape))
print("Size of ndaaray={}".format(a1.size))
```
