```python
import csv
import numpy as np
def readcsvdata():
    try:
        with open("R:\\python with KVR\\CSV\\Teacher.csv", "r") as fp:
            csvr = csv.reader(fp)
            # Skip header
            next(csvr)
            # Store CSV records
            data = list(csvr)
            print("Original CSV Data")
            for record in data:
                for val in record:
                    print("\t{}".format(val), end="\t")
                print()
            # NumPy structure
            dt = [
                ("TNO", "U8"),
                ("TNAME", "U20"),
                ("SUB", "U20")
            ]

            # Create structured NumPy array
            a = np.array([tuple(record) for record in data], dtype=dt)
            print("\nOriginal Data")
            for record in a:
                print(record)
            print("-" * 50)
            # Sort by TNO descending
            a = np.sort(a, order="TNAME")[::-1]
            print("Sorted Data Based on TNAME --> Descending Order")
            for record in a:
                print(record)

            print("-" * 50)
    except FileNotFoundError:
        print("CSV File Does Not Exist")


# Main Program
readcsvdata()
```

```python
import csv
import numpy as np

def readcsvdata():
    try:
        with open("R:\\python with KVR\\CSV\\Employee.csv", "r") as fp:

            csvr = csv.reader(fp)

            # Skip header
            next(csvr)

            # Store CSV records and ignore empty rows
            data = [record for record in csvr if record]

            print("Original CSV Data")

            for record in data:
                for val in record:
                    print("\t{}".format(val), end="\t")
                print()

            # NumPy structure
            dt = [
                ("ENO", "i4"),
                ("ENAME", "U20"),
                ("SAL", "f4")
            ]

            # Create NumPy structured array
            a = np.array([tuple(record) for record in data], dtype=dt)

            print("\nOriginal Data")

            for record in a:
                print(record)

            print("-" * 50)

            # Sort by ENAME - descending
            a = np.sort(a, order="ENAME")[::-1]

            print("Sorted Data Based on ENAME --> Descending Order")

            for record in a:
                print(record)

            print("-" * 50)

    except FileNotFoundError:
        print("CSV File Does Not Exist")


# Main Program
readcsvdata()
```
