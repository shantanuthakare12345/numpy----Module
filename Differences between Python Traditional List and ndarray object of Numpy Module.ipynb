```python
#vector Based Operations
lst=[10,20,30,40,50]
print("Content of lst={}  type={}".format(lst,type(lst)))
#Increse every value of lst by 1
lst = lst+1  # vector based operations--bot possible traditional list
```

```python
#Vector Based Operations Not Possible By Default-
#Then How do u Get It on List
#Vector Based Operations
lst=[10,20,30,40]
print("content of lst={}  type={}".format(lst,type(lst)))
for i in range(len(lst)):
    lst[i] = lst[i]+1
print("content of lst={}  type={}".format(lst,type(lst)))
```

```python
#Vector Based Operations By using  ndarray Object
lst=[10,20,30,40]
print("content of lst={}  type={}".format(lst,type(lst)))
a = np.array(lst)
print("content of a={}  type={}".format(a,type(a)))
a=a+1
print("content of a={}  type={}".format(a,type(a)))
```
