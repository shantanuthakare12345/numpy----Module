```python
# uniform()  : uniform Distribution
#---------------------------------------------------------------------------------
#this function generates random sample(s) between two intervals(low,high)
#the default random is low is 0.0 and hight is 1.0
#syntax-1:  varname=numpay.random.uniform()
#syntax-2:  varname=numpay.random.uniform(low,High)
#syntax-3:  varname=numpay.random.uniform(low,High,size)
```

```python
#Generates one random saample value between 0.0 to 1.0 with uniform()
from numpy import random as r
val=r.uniform()
print("random sample={}  type={}".format(val,type(val)))
```

```python
val=r.uniform()
print("random sample={}  \ntype={}".format(val,type(val)))
```

```python
val=r.uniform(size=5)
print("random sample={}  \ntype={}".format(val,type(val)))
```

```python
val=r.uniform(10,12)
print("random sample={}  type={}".format(val,type(val)))
```

```python
val=r.uniform(10.5,12)
print("random sample={}  \ntype={}".format(val,type(val)))
```

```python
val=r.uniform(10.5,12,5)
print("random sample={}  \ntype={}".format(val,type(val)))
```

```python
val=r.uniform(100,111,4)
print("random sample={}  \ntype={}".format(val,type(val)))
```

```python
val=r.uniform(100,111,(3,3))
print("random sample=\n{}  \ntype={}".format(val,type(val)))
```

```python
val=r.uniform(100,111,(2,2,2))
print("random sample=\n{}  \ntype={}".format(val,type(val)))
```
