```python
#Code for Building  a Matrix with dim of 2 with All entries as 0
import numpy as np
lst=[0,0,0,0]
a=np.array(lst).reshape(2,2)
print("Zerox Matrix")
print(a)
print("Dim od a=",a.ndim)
```

```python
#Create a object of ndarray with 6 zeros
import numpy as np
a=np.zeros(6)
print("Content of a={}  Type of a={}".format(a,type(a)))
print("Dim of a={}",a.ndim)
```

```python
#Create a object of ndarray with 6 zeros
import numpy as np
a=np.zeros(6,dtype=int)
print("Content of a={}  Type of a={}".format(a,type(a)))
print("Dim of a={}",a.ndim)
```

```python
#Create an object of ndarray with Dim 2 of 9 Elements as Zeros
import numpy as np
a=np.zeros((3,3))
print("Content of a=\n{}  Type of a={}".format(a,type(a)))
print("Dim of a={}",a.ndim)
```

```python
#Create an object of ndarray with Dim 2 of 9 Elements as Zeros
import numpy as np
a=np.zeros((3,3),dtype=int)
print("Content of a=\n{}  Type of a={}".format(a,type(a)))
print("Dim of a={}",a.ndim)
```
