```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("-"*50)
print("1-Darray Data={}  Dim={}".format(a,a.ndim))
print("-"*50)
for val in a:
    print("\t{}".format(val))
print("-"*50)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("-"*50)
print("2-Darray Data=\n{}  Dim={}".format(a,a.ndim))
print("-"*50)
for row in a:
    for val in row:
        print("\t{}".format(val))
    print("-"*30)
print("-"*50)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,12,15,18]).reshape(2,3,2)
print("-"*50)
print("3-Darray Data=\n{}  Dim={}".format(a,a.ndim))
print("-"*50)
for matrix in a:
    for row in matrix:
        for val in row:
            print("\t{}".format(val))
        print("-"*20)
    print("-"*35)
print("-"*50)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,12,15,18,35,45,65,25]).reshape(2,2,2,2)
print("-"*50)
print("4-Darray Data=\n{}  Dim={}".format(a,a.ndim))
print("-"*50)
for cluster in a:
    for matrix in cluster:
        for row in matrix:
            for val in row:
                print("\t{}".format(val))
            print("-"*20)
        print("-"*35)
    print("-"*50)
print("-"*60)
```

```python
#2.By Using numpy.nditer()
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("-"*50)
print("1-Darray Data={}  Dim={}".format(a,a.ndim))
print("-"*50)
for val in np.nditer(a):
    print("\t{}".format(val))
    print("-"*30)
print("-"*50)
```

```python
#3.By Using numpy.ndenumerate()
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("-"*50)
print("1-Darray Data={}  Dim={}".format(a,a.ndim))
print("-"*50)
for val in np.ndenumerate(a):
    print("\t{}".format(val))
print("-"*50)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("-"*50)
print("1-Darray Data={}  Dim={}".format(a,a.ndim))
print("-"*50)
for val in np.ndenumerate(a):
    print("\t{}--->{}".format(val[0],val[1]))
print("-"*50)
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("-"*50)
print("1-Darray Data={}  Dim={}".format(a,a.ndim))
print("-"*50)
for val in np.ndenumerate(a):
    print("\t{}--->{}".format(val[0][0],val[1]))
print("-"*50)
```
