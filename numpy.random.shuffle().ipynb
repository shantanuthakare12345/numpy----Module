```python
#syntax:  numpy.random.shuffle(ndarray)
```

```python
import numpy as np
a=np.arange(10)
print("Original Content of a={}  id={}".format(a,id(a)))
#re-arrange the element of ndarray in itself(inplace)
np.random.shuffle(a)
print("Re-arranged Content of a={}  id={}".format(a,id(a)))
```

```python
import numpy as np
a=np.arange(10)
print("Original Content of a={}  id={}".format(a,id(a)))
#re-arrange the element of ndarray in itself(inplace)
for i in range(1,6):
    np.random.shuffle(a)
    print("{} Time: Re-arranged Content of a={}  id={}".format(i,a,id(a)))
```

```python
import numpy as np
a=np.array((list("MISSISSIPPI")))
print("Original Content of a={}  \nid={}  type={}".format(a,id(a),type(a)))
print("-"*60)
#re-arrange the element of ndarray in itself(inplace)
for i in range(1,len(a)+1):
    np.random.shuffle(a)
    print("{} Time: Re-arranged Content of a={}  id={}".format(i,"".join(a),id(a)))
print("-"*60)
```
