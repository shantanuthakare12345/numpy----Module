```python
from numpy import random as r
val=r.rand()
print("Random Sample={}  type={}".format(val,type(val)))
```

```python
val=r.rand()
print("Random Sample={}  type={}".format(val,type(val)))
```

```python
#generate 5 random sample values for uniform distribution
val=r.rand(5)
print("Random Sample={}  \ntype={}".format(val,type(val)))
```

```python
#generate 10 random sample values for uniform distribution
val=r.rand(10)
print("Random Sample={}  \ntype={}".format(val,type(val)))
```

```python
#generate 2D random sample values for uniform distribution with 3 rows and 3 clos
val=r.rand(3,3)
print("Random Sample=\n{}  \ntype={}".format(val,type(val)))
```

```python
#generate 2D random sample values for uniform distribution with 3 rows and 2 clos
val=r.rand(3,2)
print("Random Sample=\n{}  \ntype={}".format(val,type(val)))
```

```python
#generate 3D random sample values for uniform distribution with 2 matrices ,3 rows and 2 clos
val=r.rand(2,3,2)
print("Random Sample=\n{}  \ntype={}".format(val,type(val)))
```
