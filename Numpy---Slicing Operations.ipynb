```python
#Numpy---Slicing Operations on 1-D
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#Get 10 20 30 40 from 1D array
b=a[0:4]
print(b,type(b))
```

```python
#Get the Element 60 70 80 90
print(a[5:])
```

```python
#Get the Elements 30,40,50,60 with -VE Slicing
print(a[-7:-3])
```

```python
#Get 10 30 50 70 
print(a[:-2:2])
```

```python
#Get 20 40 60 80 
print(a[1::2])
```

```python
#Get the 90 80 70 60 in backward direction
print(a[-1:-5:-1])
```

```python
#Get the 90 80 70 60 in backward direction
print(a[5::][::-1])
```

```python
#Get the 90 70 50 30 10 in backward direction with -ve step
print(a[::-2])
```

```python
#Get the 90 70 50 30 10 in backward direction with +ve step
print(a[::2][::-1])
```

```python
#Numpy--Slicing Operations on 2-D
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,25,35,45,55,77,88,99]).reshape(4,4)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#Get [60,70,80] from 2D array
a[1:2,1:]
```

```python
#get [60,70,80] From 2D array with -ve Indeices
a[-3:-2,-3:]
```

```python
#get [60,70,25,35] From 2D array 
a[1:3,1:3]
```
