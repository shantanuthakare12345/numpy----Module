```python
#Create an Obejct of ndarray
import numpy as np
a=np.random.randint(100,150,size=10)
print("Ndarray Obejct Data")
print(a)
```

```python
#Search for those Elements Which are Divisible by 5
b=np.where(a%5==0)
print("Content of b={}  Type={}".format(b[0],type(b[0])))
```

```python
a[b[0]]
```

```python
#Search for those Elements Which are Divisible by 5
b=np.extract(a%5==0,a)
print("Content of b={}  Type={}".format(b,type(b)))
```

```python
#Another Example
import numpy as np
arr=np.array([10,15,20,25,30])
result = np.where(arr>20)
print("Indices=",result)  #(array([3,4]),)----Indices array
print("Data=",arr[result])
```

```python
import numpy as np
arr = np.array([2,7,8,3,10])
idx=np.where(arr%2==0)  #here idx is a boolen array
print(idx)  #array[0,2,4]
print("Even Elements:",arr[idx]) #Array[2,8,10]
```

```python
nda=np.array([1000,2000,300,450,550,750,5000])
nss=np.where(nda<=1000,nda+nda*50/100,nda)
print(nss)
```

```python
#We can also use array for any of the arguments. Here we rpovide array valuse
import numpy as np
a = np.array([1,2,3,4])
x = np.array([10,20,30,40])
y = np.array([100,200,300,400])
z = np.where(a>2,x,y)
print(z)
```

```python
import numpy as np
a=np.array([1,2,3,4])
x=np.where((a>=2)&(a<=3),True,False)
print(x)
print(a[x])
```
