```python
#Creating ndarray object of ndarray
import numpy as np 
a=np.array([10,20,30,40,50,60,70,80,90])
print("Ndarray Original Object a={}  id={}".format(a,id(a)))
```

```python
#Copy the Content of ndarray obejct
b=np.copy(a)
print("Ndarray Copied Object b={} id={}".format(b,id(b)))
```

```python
#Do the changes in Original Obejct and Copied Object
a[0]=1000
b[-1]=2000
print("Ndarray Original Object a={}  id={}".format(a,id(a)))
print("Ndarray Copied Object b={} id={}".format(b,id(b)))
```

```python
#Check the Base For Copied Object
print("Base for Copied Object=",b.base)
```

```python
#Check the Base For Original Object
print("Base for Original Object=",a.base)
```

```python
#Alternatively
c=a.copy()
print("Ndarray Original Object a={}  id={}".format(a,id(a)))
print("Ndarray Copied Object b={} id={}".format(c,id(c)))
```

```python
#Do the Changes in Original Obejct and Copied Object
a[0]=11
c[0]=22
print("Ndarray Original Object a={}  id={}".format(a,id(a)))
print("Ndarray Copied Object b={} id={}".format(c,id(c)))
```
