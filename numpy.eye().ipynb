```python
#Syntax: varname=numpy.eye(N,M=None,k=0,dtype)
#Generates Unit OR Identity Matrix within Square OR Rectagle Matrix
```

```python
import numpy as np
a=np.eye(3)
print("Identity Matrix")
print(a,type(a))
print("Dim of a=",a.ndim)
```

```python
import numpy as np
a=np.eye(3,dtype=int)
print("Identity Matrix")
print(a,type(a))
print("Dim of a=",a.ndim)
```
