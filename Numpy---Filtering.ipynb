```python
#Create 1D Array with Random 12 Integer values
from numpy import random as r
a=r.randint(25,55,size=12)
print("Random Elements")
print(a)
```

```python
#Get the Factors of 5 --Normal Python Code
lst=[]  #Empty List
for val in a:
    if (val%5==0):
        lst.append(val)
print("Factor of 5=",lst)
```

```python
#Get the Factors of 5 --Normal Python Code
lst = [val for val in a if(val%5==0)]
print("Factor of 5=",lst)
```

```python
#Get the Factors of 5 
#Step-1: Prepare thr condition with ndarray object
ba=a%5==0
print("-"*50)
print("Boolen Array={}  Type={}".format(ba,type(ba)))
print("-"*50)
#Step-2: Pass the Boolean Array to the ndarray Object
print("Factors of 5")
print(a[ba])
```

```python
#Create 1D Array with Random 15 Integer values
from numpy import random as r
a=r.randint(25,55,size=12)
print("Random Elements")
print(a)
#Get the Factors of 5 and 3
print("-"*50)
print("Factors of 5 and 3")
print(a[(a%5==0)&(a%3==0)])
```
